```python
s1 = set()
print(type(s1))
s2 = {}
print(type(s2))
```

```python
s1 = {10,20,30,10,30,30,20}
print(s1)
s2 = set({10,20,30,10,20,30,40})
print(s2)
s3 = set([10,20,30,10,20,30,40]) #
print(s3)
print(type(s3))
```

```python
s1 ={1,2}
s1.add(3)
s1.update({4,5,5})
s1.update([4,6,5])
print(s1)
```

```python
s1 = {10,20,30,50,60,3}
s1.remove(10)
print(s1)
s1.discard(10)
print(s1)
print(s1.discard(10))
print(s1)
```

```python
s1.pop() # remove top of thre elsement
print(s1)
```

```python
s1.clear() # clear whole set
print(s1)
```

```python
s1 = {1,2,3}
s2={21,1,4,1,3,1,2,3,4,5}
print(s1.union(s2))
print(s1| s2)
```

```python
print(s1.intersection(s2))
print(s1 & s2)
```

```python
print(s2.difference(s1))
print(s2- s1)
print(s1-s2)
```

```python
s1 = {1,2,3}
s2={21,1,4,1,3,1,2,3,4,5}
print(s1.symmetric_difference(s2)) # It returns elements that are in either set, but NOT in both.
print(s1^s2)
print(s2^s1)
```

```python
s1 = {1, 2, 3}  # create a set with elements 1, 2, 3
s2 = {21, 1, 4, 1, 3, 1, 2, 3, 4, 5}  # duplicates will be removed automatically → {1,2,3,4,5,21}
print(s1.intersection_update(s2))  # updates s1 with common elements (intersection) and returns None
# After this operation:
# s1 becomes {1, 2, 3}  # common elements in both sets
# The print statement shows None because intersection_update() does not return the set
```

```python
s1 ={10,20,30,2,30,50,}
print(30 in s1)
print(300 not in s1)
print(30 not in s1)
```

```python
#iterate
s1 ={10,20,30,2,30,50}
for value in s1:
    print(value)
```

```python
s1 ={10,20,30,2,30,50}
list1 = list(s1)
for value in list1:
    print(value)
```

```python
s1 = {10, 20, 30, 2, 30, 50}  # duplicates removed → {2, 10, 20, 30, 50}
list1 = list(s1)  # convert set to list
for key, value in enumerate(list1):  # enumerate gives index and value
    print(key, value)
```

```python
s1 = {10, 20, 30, 2, 30, 50}  # set created → duplicates removed automatically → {2, 10, 20, 30, 50}
iterator = iter(s1)  # create an iterator object to traverse the set
print(next(iterator))  # fetch and print the first element from the iterator
print(next(iterator))  # fetch and print the next element from the iterator
```

```python
# convert set to tuple
s1 = {10, 20, 30, 30, 2}  # create a set → duplicates removed → {2, 10, 20, 30}
t1 = tuple(s1)  # convert the set into a tuple
print(t1)  # print the tuple
```
