```python
my_tuple = (10, 20, 30, 40, 50)

print(my_tuple[0])  # 10 (first element)
print(my_tuple[2])  # 30 (third element)
print(my_tuple[4])  # 50 (last element)
print(my_tuple[-1])  # 50 (last element)
print(my_tuple[-2])  # 40 (second last)
print(my_tuple[1:4])   # (20, 30, 40)
print(my_tuple[:3])    # (10, 20, 30)
print(my_tuple[2:])    # (30, 40, 50)

#my_tuple[1] = 100   # ❌ Error (TypeError)
# tuple is immutable we cant change the value 
```

```python
t1 = (10,20,50,60)
list1= list(t1)  # convert tuple to list
list1[0] = 100 # modidy here as a list  # modify element (index 0)
t1 = tuple(list1)  # convert back to tuple
print(t1)
```

```python
t = (1, 2, 3, 2, 2, 4) # manual way to do that check count of occurance

count = 0
for i in t:
    if i == 2:
        count += 1

print(count)  # 3
```

```python
# Step 1: Create tuples
t1 = (10, 20, 30)
t2 = (40, 50, 60)

# Step 2: Add (concatenate) two tuples
t3 = t1 + t2
print("After adding t1 and t2:", t3)

# Step 3: Multiply tuple by 2
t4 = t3 * 2
print("After multiplying by 2:", t4)

# Step 4: Check membership
print("Is 20 in t1?", 20 in t1)
print("Is 200 not in t1?", 200 not in t1)
```

```python
# iteration  for loop simple
t = (10, 20, 30, 40, 50)
for item in t:
    print(item)
```

```python
#Using for Loop with Index
for i in range(len(t)):
    print("Index:", i, "Value:", t[i])
```

```python
#while loop
i = 0

while i < len(t):
    print(t[i])
    i += 1
```

```python
#nested tuple
t = (1, 2, (3, 4, 5), 6)
print(t[0])   # 1
print(t[1])   # 2
print(t[2])   # (3, 4, 5)
```

```python
t = (1, 2, (3, 4, 5), 6)
for item in t:
    if isinstance(item, tuple):
        for inner in item:
            print(inner)
    else:
        print(item)
```

```python
nested_tuple = ((1,"agentic ai"),(2,"Gen AI"),(3,"NLP"))
for inner_tuple in nested_tuple:
    for ele in inner_tuple:
        print(ele)
```
