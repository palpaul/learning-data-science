```python
#unpacking example
numbers = [1, 2, 3]

a, b, c = numbers

print(a)  # 1
print(b)  # 2
print(c)  # 3
```

```python
#packing example
my_list1 = ["one","two","three"]
my_list2 = [1,2,3]
my_list3 = list(zip(my_list1,my_list2))
print(my_list3)

```

```python
print(my_list3[0])
```

```python
print(type(my_list3[0]))
print(type(my_list3))
```

```python
print(my_list2)
print(len(my_list2))
print(min(my_list2))
print(max(my_list2))
print(sum(my_list2))
```

```python
# tuple - orderd
#immutable
#allows duplicates
#represents - ()
# faster compared to list
#
t1 = ()
print(t1)
t2 = (10,20,30,40)
print(t2)
t3 = (10,10,2,"hello",True)
print(t3)
print(type(t1))
t4= (10)
print(type(t4))
t6 = (10,)
print(type(t6))
t7 = 10,30,30,40
print(type(t7))
```

```python
my_tuple = (10, 20, 30, 40, 50)

print(my_tuple[0])  # 10 (first element)
print(my_tuple[2])  # 30 (third element)
print(my_tuple[4])  # 50 (last element)
print(my_tuple[-1])  # 50 (last element)
print(my_tuple[-2])  # 40 (second last)
print(my_tuple[1:4])   # (20, 30, 40)
print(my_tuple[:3])    # (10, 20, 30)
print(my_tuple[2:])    # (30, 40, 50)

#my_tuple[1] = 100   # ❌ Error (TypeError)
# tuple is immutable we cant change the value 
```

```python
t1 = (10,20,50,60)
list1= list(t1)  # convert tuple to list
list1[0] = 100 # modidy here as a list  # modify element (index 0)
t1 = tuple(list1)  # convert back to tuple
print(t1)
```
